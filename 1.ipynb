```python
from typing import TypedDict , List 
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name : str
```

```python
def hello_node(state : AgentState) -> AgentState :
    """ simple hello function """
    state['name'] = f"hello {state['name']} how are you "
    return state
```

```python
graph = StateGraph(AgentState)

graph.set_finish_point("say hello")
graph.add_node("say hello", hello_node)
graph.set_entry_point('say hello')

app = graph.compile()
```

```python
from IPython.display import  display , Image
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
res = app.invoke({'name':'dev'})
res['name']
```
